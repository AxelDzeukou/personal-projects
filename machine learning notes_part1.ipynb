```python
notes from https://concordia.udemy.com/course/machinelearning/learn/lecture/19019942?start=0#overview

#Reinforcement learning
    #we do not split the data
    #upper confidence bound
        #pick slot machine which is left sqewed
#which is the most attractive ad for the user
        #reward is the value assigned in the dataset if the value is selected or not
        #Great question — evaluating whether UCB found the correct ad (or best option) comes down to how well it minimizes regret and converges to the optimal choice over time.

        #To measure if UCB found the best ad, ask:

❓ Do I know which ad is truly best (true CTR)?

📊 Is UCB choosing that ad more than others as time goes on?

📈 Is the overall reward improving and stabilizing?

If the answers are yes → UCB is working well and selecting the best ad.

    #thompson sampling
        #it is generally more powerful than ucb
        #number of times ad got reward 1 and 0 is counted

#Association rule learning
    #People who bought also bought ... That is what Association Rule Learning will help us figure out!

    #Apriori
        #No, we do not typically split the dataset into train and test sets when using the Apriori algorithm.
        #If you get too few rules, lower support or confidence. If you get too many weak rules, raise support or confidence. Try plotting item frequency to guide your min_support choice.

    #Eclat
        #its only support
        #✅ Apriori starts from individual items, calculates support, and combines them step by step to build larger itemsets.
         ✅ Eclat starts from item pairs and uses set intersection to find frequent itemsets.

        #🟢 Use Apriori if: your dataset is small/medium, you want to understand results clearly, or you're starting out.
        🔵 Use Eclat if: you need speed and scalability, and you can handle higher memory usage.

        #association rules are not symettrical

#gaussian models

    #| Situation                             | Better Alternative          |
| ------------------------------------- | --------------------------- |
| High-dimensional sparse data          | DBSCAN, HDBSCAN             |
| You want fast, simple clustering      | K-Means                     |
| Non-Gaussian or arbitrary-shaped data | Spectral clustering, DBSCAN |

    | Use Case                          | GMM?               |
| --------------------------------- | ------------------ |
| Elliptical / overlapping clusters | ✅                  |
| Hard clustering only              | ❌ (K-Means better) |
| Probabilistic labels needed       | ✅                  |
| Anomaly detection                 | ✅                  |
| Non-Gaussian shapes               | ❌                  |

#autoencoders
    #An autoencoder is an unsperivsided neural network used to learn efficient representations (encodings) of data, typically for dimensionality reduction, denoising, or anomaly detection.

It tries to reconstruct its input — meaning it learns how to compress data and then decompress it back.

    #| Component   | Role                                                                  |
| ----------- | --------------------------------------------------------------------- |
| **Encoder** | Learns to compress input data into a smaller, latent space (features) |
| **Decoder** | Learns to reconstruct the original data from the compressed version   |

Input → [Encoder] → Bottleneck (latent space) → [Decoder] → Reconstructed Input

    | Use Case               | Why Use Autoencoders?                        |
| ---------------------- | -------------------------------------------- |
| **Denoising images**   | Remove noise while keeping core structure    |
| **Anomaly detection**  | Anomalies reconstruct poorly → high error    |
| **Compression**        | Learn compact representations of data        |
| **Data visualization** | Reduce high-dimensional data to 2D or 3D     |
| **Pretraining**        | Learn features before supervised fine-tuning |

    #| Concept              | Description                                          |
| -------------------- | ---------------------------------------------------- |
| **Bottleneck**       | The compressed representation (features)             |
| **Loss Function**    | Usually MSE or Binary Crossentropy                   |
| **Undercomplete AE** | Bottleneck < input size (forces compression)         |
| **Overcomplete AE**  | Bottleneck ≥ input size (risks learning identity)    |
| **Denoising AE**     | Trained to reconstruct clean output from noisy input |
| **Variational AE**   | Learns a distribution (used in generative models)    |
```
